```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Dataset_Pegawai.csv')
dc=df.dropna()
dc
```

```python
dg = dc.groupby(['Education','YearsExperience'])['Salary'].sum().reset_index()
dg
```

```python
plt.figure(figsize=(10,5))

def pengalaman(tahun):
    if tahun<5:
        return '< 5 Tahun'
    elif tahun<=15:
        return '<= 15 Tahun'
    else:
        return '>= 15 Tahun' 


dg['pg']= dg['YearsExperience'].apply(pengalaman)

warna=['#2b5c8f','#d95f02','#7570b3']

sns.set_theme(style='whitegrid')

sns.barplot(data=dg,x='Education',y='Salary',hue='pg',estimator='mean',palette=warna,errorbar=None)

plt.title('Rata-Rata Gaji Berdasarkan Pengalaman dan Pendidikan',fontweight='bold',pad=15)
plt.xlabel('Pendidikan')
plt.ylabel('Rata-rata Gaji (Jt)')
plt.legend(title='Tahun Pengalaman',loc='upper right')

plt.savefig('Perbandingan_Rata-Rata_Gaji_Berdasarkan_Pengalama_dan_Pendidikan.png',dpi=300,bbox_inches='tight')
plt.show    
```
